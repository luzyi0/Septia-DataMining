```python
# Import pustaka dasar
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from google.colab import files
import warnings
warnings.filterwarnings('ignore')

# Meminta pengguna mengunggah file
print("Silakan unggah file dataset CSV Anda:")
uploaded = files.upload()

# Mendapatkan nama file secara otomatis dan membaca dataset
file_name = list(uploaded.keys())[0]
df = pd.read_csv(file_name)

print(f"\nDataset '{file_name}' berhasil dimuat!")
display(df.head())
```

```python
print("--- Info Dataset ---")
display(df.info())

print("\n--- Ringkasan Statistik Dasar ---")
display(df.describe())

# Visualisasi Matriks Korelasi (Hanya memproses kolom angka)
plt.figure(figsize=(10, 8))
korelasi = df.select_dtypes(include=np.number).corr()
sns.heatmap(korelasi, annot=True, cmap='coolwarm', fmt=".2f", linewidths=0.5)
plt.title("Heatmap Korelasi Antar Variabel Numerik")
plt.show()
```

```python
from sklearn.model_selection import train_test_split
from sklearn.impute import KNNImputer
from sklearn.preprocessing import LabelEncoder, RobustScaler

# 1. Buang kolom metadata, teks panjang, dan kolom yang terlalu banyak kosong (seperti 'title')
kolom_dibuang = ['pkey', 'date', 'time', 'url', 'image_url', 'report_dat', 'tags', 'title', 'text']
df_bersih = df.drop(columns=kolom_dibuang, errors='ignore')

# 2. Tentukan target prediksi
TARGET_COL = 'disaster_type'

# Memisahkan Fitur (X) dan Target/Label (y)
X = df_bersih.drop(columns=[TARGET_COL])
y = df_bersih[TARGET_COL]

# 3. Encoding Target (Mengubah jenis bencana teks menjadi angka)
le = LabelEncoder()
y = le.fit_transform(y)

# 4. One-Hot Encoding (Mengubah kolom lokasi seperti Desa/Kecamatan menjadi kolom angka)
X = pd.get_dummies(X, drop_first=True)

# 5. Imputasi (Mengisi nilai ID_Desa, long, lat yang mungkin kosong)
imputer = KNNImputer(n_neighbors=5)
X_imputed = pd.DataFrame(imputer.fit_transform(X), columns=X.columns)

# 6. Normalisasi Data
scaler = RobustScaler()
X_scaled = pd.DataFrame(scaler.fit_transform(X_imputed), columns=X.columns)

print("Pra-pemrosesan selesai. Data siap dipisahkan.")
```

```python
from sklearn.model_selection import train_test_split
from imblearn.over_sampling import RandomOverSampler
import pandas as pd

# Pisahkan dataset (80% latih, 20% uji)
# stratify=y memastikan proporsi jenis bencana di data uji dan latih sama
X_train, X_test, y_train, y_test = train_test_split(
    X_scaled, y, test_size=0.2, random_state=42, stratify=y
)

print("Distribusi target SEBELUM disetarakan:\n", pd.Series(y_train).value_counts())

# Menggunakan RandomOverSampler sebagai pengganti SMOTE
# Ini aman untuk dataset yang memiliki kelas minoritas sangat ekstrem (seperti target 3 yang hanya 2 baris)
ros = RandomOverSampler(random_state=42)
X_train_smote, y_train_smote = ros.fit_resample(X_train, y_train)

print("\nDistribusi target SETELAH disetarakan (Data sudah seimbang):\n", pd.Series(y_train_smote).value_counts())
```

```python
from sklearn.ensemble import RandomForestClassifier
from sklearn.model_selection import RandomizedSearchCV
import time

start_time = time.time()

# Rentang parameter yang ringan namun tetap efektif
param_dist = {
    'n_estimators': [50, 100],
    'max_depth': [None, 10, 20],
    'min_samples_split': [2, 5]
}

rf = RandomForestClassifier(random_state=42)

# Menggunakan RandomizedSearch agar komputasi di Colab lebih cepat
random_search = RandomizedSearchCV(
    estimator=rf, param_distributions=param_dist,
    n_iter=5, cv=3, scoring='accuracy', n_jobs=-1, random_state=42
)

print("Melatih model Random Forest... (Proses ini mungkin memakan waktu 1-3 menit)")
random_search.fit(X_train_smote, y_train_smote)

best_model = random_search.best_estimator_
print(f"\nSelesai dalam {(time.time() - start_time):.2f} detik!")
print("Parameter model terbaik yang ditemukan:", random_search.best_params_)
```

```python
from sklearn.metrics import classification_report, confusion_matrix, accuracy_score
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd

# Membuat prediksi terhadap data uji
y_pred = best_model.predict(X_test)

# Mengubah target dari angka kembali menjadi teks jenis bencana aslinya
# 'le' adalah LabelEncoder yang kita jalankan di Sel 3
y_test_label = le.inverse_transform(y_test)
y_pred_label = le.inverse_transform(y_pred)

print(f"Akurasi Keseluruhan: {accuracy_score(y_test, y_pred) * 100:.2f}%\n")
print("--- Laporan Klasifikasi ---")
# zero_division=0 mencegah error jika ada jenis bencana yang gagal diprediksi sama sekali
print(classification_report(y_test_label, y_pred_label, zero_division=0))

# 1. Visualisasi Confusion Matrix
cm = confusion_matrix(y_test_label, y_pred_label, labels=le.classes_)
plt.figure(figsize=(10, 8))
sns.heatmap(cm, annot=True, fmt='d', cmap='Blues',
            xticklabels=le.classes_, yticklabels=le.classes_)
plt.title('Confusion Matrix: Prediksi vs Fakta Aktual')
plt.xlabel('Prediksi Model')
plt.ylabel('Fakta Aktual')
plt.xticks(rotation=45, ha='right')
plt.show()

# 2. Fitur Paling Berpengaruh (Feature Importance)
# Mengambil 15 fitur teratas karena total kolom bisa mencapai ribuan
importances = best_model.feature_importances_
feat_imp = pd.Series(importances, index=X_scaled.columns).sort_values(ascending=False).head(15)

plt.figure(figsize=(10, 6))
sns.barplot(x=feat_imp.values, y=feat_imp.index, palette='viridis')
plt.title('15 Kolom/Fitur Paling Berpengaruh Terhadap Prediksi Bencana')
plt.xlabel('Tingkat Kepentingan')
plt.ylabel('Nama Kolom (Termasuk wilayah hasil ekstrak)')
plt.show()
```
